```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
dengue = pd.read_csv("datasets\\dengue.csv")
```

```python
dengue.info()
```

```python
dengue["Month"] = dengue["Month"].astype("category")
dengue["Year"] = dengue["Year"].astype("category")
dengue["Region"] = dengue["Region"].astype("category")
```

```python
dengue.isnull().sum()
```

<h1>Insight No. 1</h1>
<h3>In which region has the highest dengue cases?</h3>
<h2 style="color:red">Region IV-A has the highest number of dengue case</h2>

```python
sns.barplot(data=dengue, x = "Region", y = "Dengue_Cases")
plt.gcf().set_size_inches(17, 7)
plt.show()
```

<h1>Insight No. 2</h1>
<h3>In which region has the highest dengue deaths?</h3>
<h2 style="color:red">NCR has the highest number of dengue deaths</h2>

```python
case_per_region = dengue.groupby("Region")["Dengue_Deaths"].sum().reset_index()
sns.barplot(data=case_per_region, x = "Region", y = "Dengue_Deaths")
plt.title("Total Dengue Case per Region")
plt.gcf().set_size_inches(17, 7)
plt.show()
```

<h1>Insight No. 3</h1>
<h3>What is the average dengue case in each region?<</h3>
<h2 style="color:red">NCR has the highest number of dengue deaths</h2>

```python
avg_cases = dengue.groupby("Region")["Dengue_Cases"].mean().reset_index()
sns.barplot(data=avg_cases,x="Dengue_Cases", y="Region")
plt.title("Average Dengue Cases per Region")
plt.gcf().set_size_inches(17, 7)
plt.xlabel("Average Cases")
plt.ylabel("Region")
plt.show()
```

<h1>Insight No. 4</h1>
<h3>What is the average dengue deaths in each region?<</h3>
<h2 style="color:red">NCR has the highest number of dengue deaths</h2>

```python
avg_deaths = dengue.groupby("Region")["Dengue_Deaths"].mean().reset_index()
sns.barplot(data=avg_deaths, x="Dengue_Deaths", y="Region")
plt.title("Average Dengue Deaths per Region")
plt.xlabel("Average Deaths")
plt.ylabel("Region")
plt.show()
```

<h1>Insight No. 7</h1>
<h3>In which month has the highest dengue deaths of all time?</h3>
<h2 style="color:red">Month of October has the highest number of dengue deaths of all time</h2>

```python
order = dengue["Month"].unique()
case_per_month = dengue.groupby("Month")["Dengue_Deaths"].sum().reset_index()
sns.barplot(data=case_per_month, x = "Month", y = "Dengue_Deaths", order=order)
plt.title("Total Dengue Deaths Per Month")
plt.gcf().set_size_inches(12, 5)
sns.barplot(data=dengue, x = "Month", y = "Dengue_Deaths", order=order)
plt.gcf().set_size_inches(14, 7)
plt.show()
```

<h1>Insight No. 10</h1>
<h3>Heatmap of the dengue cases by year and month</h3>
<h2 style="color:red">It shows Auguest and September in 2019 has the highest number of cases</h2>

```python
heatmap_data = dengue.pivot_table(values="Dengue_Cases", index="Month", columns="Year", aggfunc="sum")

sns.heatmap(heatmap_data, annot=True, fmt=".0f")
plt.title("Heatmap of Dengue Cases by Year and Month")
plt.xlabel("Year")
plt.ylabel("Month")
plt.show()
```

<h1>Insight No. 11</h1>
<h3>Total Cases vs Total Deaths over the 5 years span</h3>
<h2 style="color:red">It shows many cases over the years but now a high number of deaths</h2>

```python
total = dengue.groupby("Year")[["Dengue_Cases", "Dengue_Deaths"]].sum().reset_index()

sns.lineplot(data=total, x="Year", y="Dengue_Cases", label="Cases", marker="o")
sns.lineplot(data=total, x="Year", y="Dengue_Deaths", label="Deaths", marker="o", color="red")
plt.title("Total Dengue Cases vs Deaths Over Years")
plt.legend()
plt.show()
```

<h1>Insight No. 12</h1>
<h3>What is the average cases per month</h3>
<h2 style="color:red">It shows the average case per month, August and September being the highest</h2>

```python
monthly_avg = dengue.groupby("Month")["Dengue_Cases"].mean().reset_index()

sns.barplot(data=monthly_avg, x="Month", y="Dengue_Cases", palette="crest", order=order)
plt.title("Average Dengue Cases per Month")
plt.xticks(rotation=45)
plt.show()
```

<h1>Insight No. 13</h1>
<h3>How many regions has over 5000 cases per year?</h3>
<h2 style="color:red">There are 16 regions that has over 5000 cases of dengue for the year 2019</h2>

```python
region_per_year = dengue.groupby(["Year", "Region"])["Dengue_Cases"].sum().reset_index()
case_over_5k = region_per_year[region_per_year["Dengue_Cases"] > 5000]
region_counts = case_over_5k.groupby("Year").size().reset_index(name="regions_over_5k")
sns.barplot(data=region_counts, x="Year", y="regions_over_5k", palette="Set3")
plt.title("Regions Exceeding 5,000 Cases per Year")
plt.ylabel("Number of Regions")
plt.xlabel("Year")
plt.show()
```

<h1>Insight No. 14</h1>
<h2 style="color:red">Average Dengue Cases by Month</h2>

```python
month_order = ["January", "February", "March", "April", "May", "June", 
               "July", "August", "September", "October", "November", "December"]

monthly_avg = dengue.groupby("Month")["Dengue_Cases"].mean().reindex(month_order).reset_index()

sns.barplot(data=monthly_avg, x="Month", y="Dengue_Cases", palette="coolwarm")
plt.title("Average Dengue Cases per Month")
plt.xticks(rotation=45)
plt.ylabel("Average Cases")
plt.xlabel("Month")
plt.show()
```

<h1>Insight No. 15</h1>
<h2 style="color:red">Case Fatality Rate (CFR) by Region</h2>

```python
region_stats = dengue.groupby("Region")[["Dengue_Cases", "Dengue_Deaths"]].sum().reset_index()
region_stats["CFR (%)"] = (region_stats["Dengue_Deaths"] / region_stats["Dengue_Cases"]) * 100
region_stats = region_stats.sort_values("CFR (%)", ascending=False)

sns.barplot(data=region_stats, x="CFR (%)", y="Region", palette="Reds")
plt.title("Case Fatality Rate (%) by Region")
plt.xlabel("CFR (%)")
plt.ylabel("Region")
plt.show()
```

<h1>Insight No. 16</h1>
<h2 style="color:red">Correlation Between Dengue Cases and Deaths</h2>

```python
sns.scatterplot(data=dengue, x="Dengue_Cases", y="Dengue_Deaths", hue="Region", alpha=0.6)
plt.title("Correlation between Cases and Deaths")
plt.xlabel("Dengue Cases")
plt.ylabel("Dengue Deaths")
plt.show()

correlation = dengue["Dengue_Cases"].corr(dengue["Dengue_Deaths"])
print("Correlation coefficient between Cases and Deaths:", correlation)
```

<h1>Insight No. 17</h1>
<h2 style="color:red">Regions with Dengue Deaths Over 50 in a Year
</h2>

```python
deaths_by_region_year = dengue.groupby(["Year", "Region"])["Dengue_Deaths"].sum().reset_index()
high_death_regions = deaths_by_region_year[deaths_by_region_year["Dengue_Deaths"] > 50]

sns.barplot(data=high_death_regions, x="Year", y="Dengue_Deaths", hue="Region", palette="Set2")
plt.title("Regions with Dengue Deaths over 50 per Year")
plt.ylabel("Dengue Deaths")
plt.xlabel("Year")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

<h1>Insight No. 18</h1>
<h2 style="color:red">Year with the Highest total death </h2>

```python
year_deaths = dengue.groupby("Year")["Dengue_Deaths"].sum().reset_index()

sns.barplot(data=year_deaths, x="Year", y="Dengue_Deaths", palette="Reds")
plt.title("Total Dengue Deaths per Year")
plt.xlabel("Year")
plt.ylabel("Deaths")
plt.xticks(rotation=45)
plt.show()
```

<h1>Insight No. 19</h1>
<h2 style="color:red">Number of Regions with Over 10,000 Cases per Year</h2>

```python
region_per_year = dengue.groupby(["Year", "Region"])["Dengue_Cases"].sum().reset_index()
high_cases = region_per_year[region_per_year["Dengue_Cases"] > 10000]
region_counts = high_cases.groupby("Year").size().reset_index(name="Regions > 10k")

sns.barplot(data=region_counts, x="Year", y="Regions > 10k", palette="Set3")
plt.title("Regions with Over 10,000 Cases per Year")
plt.ylabel("Number of Regions")
plt.xlabel("Year")
plt.show()
```
