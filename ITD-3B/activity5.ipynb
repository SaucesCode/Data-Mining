```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

<h1>DENGUE DATA ANALYSIS</h1>
<h2>INCLUSIVE YEAR 2016 - 2020</h2>
<h3>Analyst: James Patrick I. De Mesa</h3>

```python
dengue = pd.read_csv("datasets\\dengue.csv")
```

```python
dengue
```

```python
dengue.info()
```

```python
dengue["Month"] = dengue["Month"].astype("category")
dengue["Year"] = dengue["Year"].astype("category")
dengue["Region"] = dengue["Region"].astype("category")
```

```python
dengue.isnull().sum()
```

<h1>Insight No. 1</h1>
<h3>In which region has the highest dengue cases?</h3>
<h2 style="color:red">Region IV-A has the highest number of dengue case</h2>

```python
sns.barplot(data=dengue, x = "Region", y = "Dengue_Cases")
plt.gcf().set_size_inches(17, 7)
plt.show()
```

<h1>Insight No. 2</h1>
<h3>In which region has the highest dengue deaths?</h3>
<h2 style="color:red">NCR has the highest number of dengue deaths</h2>

```python
case_per_region = dengue.groupby("Region")["Dengue_Deaths"].sum().reset_index()
sns.barplot(data=case_per_region, x = "Region", y = "Dengue_Deaths")
plt.title("Total Dengue Case per Region")
plt.gcf().set_size_inches(17, 7)
plt.show()
```

<h1>Insight No. 3</h1>
<h3>What is the average dengue case in each region?<</h3>
<h2 style="color:red">NCR has the highest number of dengue deaths</h2>

```python
avg_cases = dengue.groupby("Region")["Dengue_Cases"].mean().reset_index()
sns.barplot(data=avg_cases,x="Dengue_Cases", y="Region")
plt.title("Average Dengue Cases per Region")
plt.gcf().set_size_inches(17, 7)
plt.xlabel("Average Cases")
plt.ylabel("Region")
plt.show()
```

<h1>Insight No. 4</h1>
<h3>What is the average dengue deaths in each region?<</h3>
<h2 style="color:red">NCR has the highest number of dengue deaths</h2>

```python
avg_deaths = dengue.groupby("Region")["Dengue_Deaths"].mean().reset_index()
sns.barplot(data=avg_deaths, x="Dengue_Deaths", y="Region")
plt.title("Average Dengue Deaths per Region")
plt.xlabel("Average Deaths")
plt.ylabel("Region")
plt.show()
```

<h1>Insight No. 5</h1>
<h3>In which year has the highest dengue cases?</h3>
<h2 style="color:red">2019 has the highest number of dengue cases</h2>

```python
cases_by_year = dengue.groupby("Year")["Dengue_Cases"].sum().reset_index()
sns.lineplot(data=cases_by_year, x="Year", y="Dengue_Cases")
plt.title("Dengue Cases Per Year")
plt.show()
```

<h1>Insight No. 6</h1>
<h3>In which year has the highest dengue death?</h3>
<h2 style="color:red">2016 has the highest number of dengue deaths</h2>

```python
deaths_by_year = dengue.groupby("Year")["Dengue_Deaths"].sum().reset_index()
sns.lineplot(data=deaths_by_year, x="Year", y="Dengue_Deaths", marker="o", color="red")
plt.title("Dengue Deaths Per Year")
plt.show()
```

<h1>Insight No. 7</h1>
<h3>In which month has the highest dengue deaths of all time?</h3>
<h2 style="color:red">Month of October has the highest number of dengue deaths of all time</h2>

```python
order = dengue["Month"].unique()
case_per_month = dengue.groupby("Month")["Dengue_Deaths"].sum().reset_index()
sns.barplot(data=case_per_month, x = "Month", y = "Dengue_Deaths", order=order)
plt.title("Total Dengue Deaths Per Month")
plt.gcf().set_size_inches(12, 5)
sns.barplot(data=dengue, x = "Month", y = "Dengue_Deaths", order=order)
plt.gcf().set_size_inches(14, 7)
plt.show()
```

<h1>Insight No. 8</h1>
<h3>What is the top 5 months with the highest number of cases for all time?</h3>
<h2 style="color:red">The top 5 months for the highest number of cases are:</h2>
<ul>
    <li>August</li>
    <li>September</li>
    <li>July</li>
    <li>October</li>
    <li>November</li>
</ul>

```python
case_per_month = dengue.groupby("Month")["Dengue_Cases"].sum().reset_index()
top_5_months = case_per_month.nlargest(5, "Dengue_Cases")

sns.barplot(data=top_5_months, x="", y="Dengue_Cases")
plt.title("Top 5 Months with Most Dengue Cases")
plt.xlabel("Total Cases")
plt.ylabel("Month")
plt.gcf().set_size_inches(14, 7)
plt.show()
```

<h1>Insight No. 9</h1>
<h3>What is the top 5 Regions with the highest number of cases for all time?</h3>
<h2 style="color:red">The top 5 regions for the highest number of cases are:</h2>
<ul>

    <li>Region III</li>

```python
case_per_month = dengue.groupby("Region")["Dengue_Cases"].sum().reset_index()
top_5_months = case_per_month.nlargest(5, "Dengue_Cases")

sns.barplot(data=top_5_months, x="Dengue_Cases", y="Region")
plt.title("Top 5 Months with Most Dengue Cases")
plt.xlabel("Total Cases")
plt.ylabel("Month")
plt.show()
```

<h1>Insight No. 10</h1>
<h3>Heatmap of the dengue cases by year and month</h3>
<h2 style="color:red">It shows Auguest and September in 2019 has the highest number of cases</h2>

```python
heatmap_data = dengue.pivot_table(values="Dengue_Cases", index="Month", columns="Year", aggfunc="sum")

sns.heatmap(heatmap_data, annot=True, fmt=".0f")
plt.title("Heatmap of Dengue Cases by Year and Month")
plt.xlabel("Year")
plt.ylabel("Month")
plt.show()
```

<h1>Insight No. 11</h1>
<h3>Total Cases vs Total Deaths over the 5 years span</h3>
<h2 style="color:red">It shows many cases over the years but now a high number of deaths</h2>

```python
total = dengue.groupby("Year")[["Dengue_Cases", "Dengue_Deaths"]].sum().reset_index()

sns.lineplot(data=total, x="Year", y="Dengue_Cases", label="Cases", marker="o")
sns.lineplot(data=total, x="Year", y="Dengue_Deaths", label="Deaths", marker="o", color="red")
plt.title("Total Dengue Cases vs Deaths Over Years")
plt.legend()
plt.show()
```

<h1>Insight No. 12</h1>
<h3>What is the average cases per month</h3>
<h2 style="color:red">It shows the average case per month, August and September being the highest</h2>

```python
monthly_avg = dengue.groupby("Month")["Dengue_Cases"].mean().reset_index()

sns.barplot(data=monthly_avg, x="Month", y="Dengue_Cases", palette="crest", order=order)
plt.title("Average Dengue Cases per Month")
plt.xticks(rotation=45)
plt.show()
```
