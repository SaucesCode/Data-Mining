```python
import numpy as np
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt
import folium
from folium.plugins import HeatMap
import warnings 
warnings.filterwarnings('ignore')
```

```python


# Load the dataset with the correct encoding
df = pd.read_csv('datasets\\Global_terrorist.csv', encoding='ISO-8859-1')

```

<h1>DATA AGGREGATION</h1>

```python
middle_east_and_north_america = df[df['region_txt'] == "Middle East & North Africa"]
meNna_2010 = middle_east_and_north_america[middle_east_and_north_america['iyear'] > 2010]
meNna_2010_attack_type = meNna_2010[meNna_2010['attacktype1_txt'] == "Bombing/Explosion"]
final_filter = meNna_2010_attack_type.groupby(["latitude","longitude"]).size().reset_index(name="incident_count")
```

```python
meNna_2010['attacktype1_txt'].value_counts()
```

```python
heatmap_data = final_filter[['latitude', 'longitude','incident_count']].values.tolist()
```

```python
my_map = folium.Map(location = [30, 40], zoom_start = 4)
```

```python
HeatMap(heatmap_data).add_to(my_map)
```

```python
my_map
```

```python
ph = df[df['country_txt'] == "Philippines"]
ph_2000 = ph[ph['iyear'] > 1970]
ph_2000_attack_type = ph_2000[ph_2000['attacktype1_txt'] == "Hostage Taking (Kidnapping)"]
final_filter_ph = ph_2000_attack_type.groupby(["latitude","longitude"]).size().reset_index(name="incident_count")
```

```python
heatmap_data_ph = final_filter_ph[['latitude', 'longitude','incident_count']].values.tolist()
```

```python
my_map_ph = folium.Map(location = [30, 40], zoom_start = 2)
HeatMap(heatmap_data_ph).add_to(my_map_ph)
my_map_ph
```
