```python
import numpy as np
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt
import folium
from folium.plugins import HeatMap
import warnings 
warnings.filterwarnings('ignore')
```

```python


# Load the dataset with the correct encoding
df = pd.read_csv('datasets\\Global_terrorist.csv', encoding='ISO-8859-1')

```

```python
df
```

```python
df.info()
```

<h1>DATA AGGREGATION</h1>

```python
middle_east_and_north_america = df[df['region_txt'] == "Middle East & North Africa"]
meNna_2010 = middle_east_and_north_america[middle_east_and_north_america['iyear'] > 2010]
meNna_2010_attack_type = meNna_2010[meNna_2010['attacktype1_txt'] == "Bombing/Explosion"]
final_filter = meNna_2010_attack_type.groupby(["latitude","longitude"]).size().reset_index(name="incident_count")
```

```python
meNna_2010['attacktype1_txt'].value_counts()
```

```python
heatmap_data = final_filter[['latitude', 'longitude','incident_count']].values.tolist()
```

```python
my_map = folium.Map(location = [30, 40], zoom_start = 4)
```

```python
HeatMap(heatmap_data).add_to(my_map)
```

```python
my_map
```

<h1># 2</h1>

```python
ph = df[df['country_txt'] == "Philippines"]
ph_2000 = ph[ph['iyear'] > 1970]
ph_2000_attack_type = ph_2000[ph_2000['attacktype1_txt'] == "Hostage Taking (Kidnapping)"]
final_filter_ph = ph_2000_attack_type.groupby(["latitude","longitude"]).size().reset_index(name="incident_count")
```

<h1># 3</h1>

```python
target = df[df['targtype1_txt'] == "Private Citizens & Property"] #43k
target_filter1 = target[target['attacktype1_txt'] == "Armed Assault"] #10k
target_filter2 = target_filter1[target_filter1['weaptype1_txt'] == 'Firearms'] #8k
target_filter3 = target_filter2[target_filter2['region_txt'] == "North America"] #113
#final filter
target_filterf = target_filter3.groupby(["latitude","longitude"]).size().reset_index(name="incident_count")
# change to list
heatmap_data_target = target_filterf[['latitude', 'longitude','incident_count']].values.tolist()
# creating map
my_map_target = folium.Map(location = [30, 40], zoom_start = 2)
HeatMap(heatmap_data_target).add_to(my_map_target)
my_map_target
```

<h1># 4</h1>

```python
region_sa = df[df['region_txt'] == "South Asia"] # 44974 
region_sa_50 = region_sa[region_sa['nkill'] + region_sa['nwound'] > 50]
region_saf = region_sa_50.groupby(["latitude", "longitude"]).size().reset_index(name="incident_count")
final_filter_sa = region_saf[region_saf['incident_count'] < 1000] # 331 

heatmap_data_sa = final_filter_sa[['latitude', 'longitude', 'incident_count']].values.tolist()

my_map_sa = folium.Map(location=[20, 77], zoom_start=4)
HeatMap(heatmap_data_sa).add_to(my_map_sa)
my_map_sa
```
