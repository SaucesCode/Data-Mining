```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
birds = pd.read_csv('datasets\\penguins.csv')
```

```python
birds
```

```python
birds.describe()
```

```python
birds.isnull().sum()
```

```python
birds['bill_length_mm'] = birds['bill_length_mm'].fillna(birds['bill_length_mm'].mean())
birds['bill_depth_mm'] = birds['bill_depth_mm'].fillna(birds['bill_depth_mm'].mean())
birds['flipper_length_mm'] = birds['flipper_length_mm'].fillna(birds['flipper_length_mm'].mean())
birds['body_mass_g'] = birds['body_mass_g'].fillna(birds['body_mass_g'].mean())
birds['sex'] = birds['sex'].fillna(birds['sex'].mode()[0])
```

```python
birds.info()
```

```python
birds['species'] = birds['species'].astype('category')
birds['island'] = birds['island'].astype('category')
birds['sex'] = birds['sex'].astype('category')
birds['year'] = birds['year'].astype('category')
```

```python
birds['weight_kilo'] = np.round(birds['body_mass_g'] / 1000, 2)
```

```python
birds[(birds['species'] == "Gentoo") & (birds['bill_depth_mm'] == birds['bill_depth_mm'].mean())]
```

```python
max_avg_blength = birds.groupby('species')['bill_length_mm'].mean().idxmax()
max_avg_bdepth = birds.groupby('species')['bill_depth_mm'].mean().idxmax()
max_avg_flength = birds.groupby('species')['flipper_length_mm'].mean().idxmax()
max_avg_bodyg = birds.groupby('species')['body_mass_g'].mean().idxmax()

print(max_avg_blength, max_avg_bdepth, max_avg_flength, max_avg_bodyg)
```
